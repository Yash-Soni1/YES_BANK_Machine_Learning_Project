# Project Summary

The YES BANK Stock Closing Price Prediction project focuses on analyzing historical stock market data to understand the movement and behavior of YES BANK share prices. Stock market prices are influenced by several factors, including market conditions, investor confidence, company performance, economic developments, and unexpected events. Understanding historical price movements can help identify patterns that may be useful for developing predictive machine learning models.

YES BANK has experienced several significant changes in its stock price over the years. The bank faced major financial and governance-related challenges, particularly around 2018, which affected investor confidence and its market performance. Studying historical stock price data provides an opportunity to understand how the stock behaved during different periods and how major fluctuations are reflected in its historical price movements.

The dataset used in this project contains monthly historical stock market information, including Date, Open, High, Low, and Close prices. The Close column represents the closing stock price and is the primary target variable for the subsequent regression task. The other numerical columns provide information about the stock's opening price and the highest and lowest prices recorded during each month.

The first stage of the project involves Exploratory Data Analysis (EDA). The dataset is examined to understand its structure, dimensions, data types, statistical properties, missing values, and duplicate records. Data quality checks are performed to identify inconsistencies and ensure that the dataset is suitable for further analysis. The Date column is converted into a datetime format, and the records are arranged chronologically.

Data wrangling and feature exploration are performed to derive additional information from the existing variables. These include monthly price changes, monthly percentage returns, price ranges, and moving averages. Since the dataset contains monthly observations, a seven-month moving average is used to examine longer-term price trends instead of a seven-day moving average.

Different visualizations are created to study the distribution of stock prices, changes over time, relationships between Open, High, Low, and Close prices, yearly trends, and monthly volatility. Univariate analysis helps understand individual variables, bivariate analysis examines relationships between two variables, and multivariate analysis explores relationships among multiple numerical variables.

Correlation analysis is also performed to understand how the numerical variables are related to one another. Outlier identification is conducted using statistical techniques such as the Interquartile Range method. However, unusual stock price movements are not automatically removed because they may represent genuine market events and important changes in investor behavior.

The insights obtained from this EDA stage will provide a foundation for the next stage of the project, which involves developing regression models to estimate future closing prices. The overall aim is to understand historical stock behavior, prepare reliable data, and identify useful features for machine learning.

This project demonstrates how data analysis and visualization can be applied to financial time-series data. The findings should be interpreted as historical patterns rather than guarantees of future stock market performance.

The analysis aims to:

The project is intended for historical data analysis and educational prediction purposes. It is not intended to provide investment advice.

```python
# YES BANK STOCK PRICE PREDICTION - COMPLETE EDA

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from google.colab import files

warnings.filterwarnings('ignore')
sns.set_style('whitegrid')

# 2. UPLOAD AND LOAD DATASET
uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("FIRST 5 ROWS")
display(df.head())

print("\nLAST 5 ROWS")
display(df.tail())

print("\nDATASET SHAPE:", df.shape)

print("\nDATASET INFORMATION")
df.info()

print("\nSTATISTICAL SUMMARY")
display(df.describe())

# 3. DATA CLEANING AND INITIAL CHECKS
print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE ROWS:", df.duplicated().sum())

# Missing Value Heatmap
plt.figure(figsize=(8, 4))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values")
plt.show()

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nUNIQUE VALUES")
print(df.nunique())

# 4. DATE CONVERSION AND SORTING
df['Date'] = pd.to_datetime(
    df['Date'],
    format='%b-%y',
    errors='coerce'
)

if df['Date'].isnull().any():
    print("\nSome dates could not be converted. Check the original date format.")
    display(df[df['Date'].isnull()])

df = df.dropna(subset=['Date'])
df = df.sort_values('Date').reset_index(drop=True)

# 5. CHECK OHLC DATA
print("\nNON-POSITIVE VALUES")
print((df[['Open', 'High', 'Low', 'Close']] <= 0).sum())

print("\nOHLC CONSISTENCY CHECK")
print(
    "High lower than Open or Close:",
    (df['High'] < df[['Open', 'Close']].max(axis=1)).sum()
)

print(
    "Low higher than Open or Close:",
    (df['Low'] > df[['Open', 'Close']].min(axis=1)).sum()
)

# 6. FEATURE ENGINEERING
df['Monthly_Change'] = df['Close'].diff()
df['Monthly_Return'] = df['Close'].pct_change() * 100
df['Price_Range'] = df['High'] - df['Low']
df['Open_Close_Difference'] = df['Close'] - df['Open']
df['MA_7'] = df['Close'].rolling(window=7).mean()
df['Rolling_Volatility'] = df['Monthly_Return'].rolling(window=7).std()
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

print("\nFEATURE-ENGINEERED DATA")
display(df.head(10))

# 7. ORGANIZED VISUALIZATION

fig, axes = plt.subplots(7, 3, figsize=(22, 35))
axes = axes.flatten()

# Chart 1: Closing Price Over Time
axes[0].plot(df['Date'], df['Close'])
axes[0].set_title("Closing Price Over Time")
axes[0].set_xlabel("Date")
axes[0].set_ylabel("Closing Price")
axes[0].tick_params(axis='x', rotation=45)

# Chart 2: Closing Price Distribution
sns.histplot(df['Close'], kde=True, ax=axes[1])
axes[1].set_title("Closing Price Distribution")

# Chart 3: Closing Price Boxplot
sns.boxplot(x=df['Close'], ax=axes[2])
axes[2].set_title("Closing Price Boxplot")

# Chart 4: Monthly Returns
axes[3].plot(df['Date'], df['Monthly_Return'])
axes[3].set_title("Monthly Returns")
axes[3].set_xlabel("Date")
axes[3].set_ylabel("Return (%)")
axes[3].tick_params(axis='x', rotation=45)

# Chart 5: Monthly Change
axes[4].plot(df['Date'], df['Monthly_Change'])
axes[4].set_title("Monthly Change")
axes[4].set_xlabel("Date")

# Chart 6: Monthly Return Distribution
sns.histplot(df['Monthly_Return'].dropna(), kde=True, ax=axes[5])
axes[5].set_title("Monthly Return Distribution")

# Chart 7: Price Range Distribution
sns.histplot(df['Price_Range'], kde=True, ax=axes[6])
axes[6].set_title("Price Range Distribution")

# Chart 8: Open vs Close
sns.scatterplot(data=df, x='Open', y='Close', ax=axes[7])
axes[7].set_title("Open vs Close")

# Chart 9: High vs Close
sns.scatterplot(data=df, x='High', y='Close', ax=axes[8])
axes[8].set_title("High vs Close")

# Chart 10: Low vs Close
sns.scatterplot(data=df, x='Low', y='Close', ax=axes[9])
axes[9].set_title("Low vs Close")

# Chart 11: High vs Low
sns.scatterplot(data=df, x='High', y='Low', ax=axes[10])
axes[10].set_title("High vs Low")

# Chart 12: Price Range vs Monthly Return
sns.scatterplot(
    data=df,
    x='Price_Range',
    y='Monthly_Return',
    ax=axes[11]
)
axes[11].set_title("Price Range vs Monthly Return")

# Chart 13: Open-Close Difference
sns.histplot(df['Open_Close_Difference'], kde=True, ax=axes[12])
axes[12].set_title("Open-Close Difference")

# Chart 14: Closing Price + Moving Average
axes[13].plot(df['Date'], df['Close'], label='Closing Price')
axes[13].plot(df['Date'], df['MA_7'], label='7-Month MA')
axes[13].set_title("Closing Price + Moving Average")
axes[13].legend()
axes[13].tick_params(axis='x', rotation=45)

# Chart 15: Rolling Volatility
axes[14].plot(df['Date'], df['Rolling_Volatility'])
axes[14].set_title("7-Month Rolling Volatility")
axes[14].tick_params(axis='x', rotation=45)

# Chart 16: Yearly Average Closing Price
df.groupby('Year')['Close'].mean().plot(
    kind='bar',
    ax=axes[15]
)
axes[15].set_title("Yearly Average Close")
axes[15].set_xlabel("Year")
axes[15].tick_params(axis='x', rotation=45)

# Chart 17: Yearly Minimum and Maximum Closing Price
df.groupby('Year')['Close'].agg(['min', 'max']).plot(
    kind='bar',
    ax=axes[16]
)
axes[16].set_title("Yearly Min and Max Close")
axes[16].set_xlabel("Year")
axes[16].tick_params(axis='x', rotation=45)

# Chart 18: Average Closing Price by Month
df.groupby('Month')['Close'].mean().plot(
    kind='bar',
    ax=axes[17]
)
axes[17].set_title("Average Close by Month")
axes[17].set_xlabel("Month")

# Chart 19: Correlation Heatmap
correlation_columns = [
    'Open', 'High', 'Low', 'Close',
    'Monthly_Change', 'Monthly_Return',
    'Price_Range', 'Open_Close_Difference',
    'MA_7', 'Rolling_Volatility'
]

sns.heatmap(
    df[correlation_columns].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.1f',
    ax=axes[18]
)
axes[18].set_title("Correlation Heatmap")

# Chart 20: Outlier Detection
sns.boxplot(
    data=df[['Open', 'High', 'Low', 'Close']],
    ax=axes[19]
)
axes[19].set_title("Outlier Detection in OHLC Prices")

# Hide Unused Subplot
axes[20].axis('off')

plt.tight_layout(pad=3)
plt.show()

# 8. PAIRPLOT

sns.pairplot(
    df[['Open', 'High', 'Low', 'Close']],
    diag_kind='hist'
)
plt.suptitle("Pairplot of Stock Prices", y=1.02)
plt.show()

# 9. OUTLIER DETECTION USING IQR

print("\nOUTLIER DETECTION USING IQR")

for column in ['Open', 'High', 'Low', 'Close']:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(f"\n{column}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))

# Outliers are identified but not automatically removed.

# 10. EDA SUMMARY

print("\n========== EDA SUMMARY ==========")

print("Total Records:", len(df))
print("Total Features:", len(df.columns))
print("Start Date:", df['Date'].min())
print("End Date:", df['Date'].max())

print("\nCLOSING PRICE STATISTICS")
print(df['Close'].describe())

print("\nMONTHLY RETURN STATISTICS")
print(df['Monthly_Return'].describe())

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE ROWS:", df.duplicated().sum())

# 11. CONCLUSION

print("\n========== CONCLUSION ==========")

print("EDA completed successfully.")
print("The dataset contains historical YES BANK stock prices.")
print("Closing Price is the target variable.")
print("Time-series trends, distributions and correlations were analyzed.")
print("A 7-month moving average was used because the dataset is monthly.")
print("Outliers were identified but not automatically removed.")
print("The dataset is ready for further preprocessing and model development.")

# 12. SAVE PROCESSED DATASET

output_file = "YES_BANK_EDA_Processed.csv"

df.to_csv(output_file, index=False)

print("\nProcessed dataset saved as:", output_file)

files.download(output_file)
```
